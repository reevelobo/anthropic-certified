## Notebook Goal
Use assistant prefilling plus stop sequences to produce only three short AWS CLI commands in a single response with no extra commentary.

````python
# Import

from anthropic import Anthropic
from dotenv import load_dotenv

# Env var
load_dotenv()
````

## Setup Notes
This cell loads your API credentials from environment variables and initializes the Anthropic SDK dependencies.

````python
# Define model and client
model = "claude-haiku-4-5-20251001"

client = Anthropic()
````

## Conversation State
Message history is stored as a list so each turn can be appended and sent back to Claude in order.

````python
## Make an initial list of messages
messages = []
````

## Helper Functions
These helpers manage chat message entries and call the Claude API with optional system prompts and stop sequences.

````python
## Define helper functions
def add_user_message(messages, text):
    return messages.append({"role": "user", "content": text})

def add_assistant_message(messages, text):
    return messages.append({"role": "assistant", "content": text})

def chat(messages, system=None, stop_sequences=None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
    }

    if system:
        params["system"] = system

    if stop_sequences:
        params["stop_sequences"] = stop_sequences

    message = client.messages.create(**params)
    return message.content[0].text
````
