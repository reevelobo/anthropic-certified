```python
# Import

from anthropic import Anthropic
from dotenv import load_dotenv

# Env var
load_dotenv()
```

```python
# Define modle and client
model = "claude-haiku-4-5-20251001"

client = Anthropic()
```

```python
## Make an initial list of messages
messages = []
```

```python
## Define helper functions
def add_user_message(messages, text):
    return messages.append({"role": "user", "content": text})

def add_assistant_message(messages, text):
    return messages.append({"role": "assistant", "content": text})

def chat(messages, system=None, temperature=None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
    }

    if system:
        params["system"] = system

    if temperature is not None:
        try:
            message = client.messages.create(**params, temperature=temperature)
            return message.content[0].text
        except TypeError as exc:
            # Some SDK/API combinations do not support temperature for this method.
            if "unexpected keyword argument 'temperature'" not in str(exc):
                raise

    message = client.messages.create(**params)
    return message.content[0].text
```

## Temperature Comparison Demo

This final demo runs the same prompt three ways:
1. Without temperature (API default)
2. With low temperature (`0.0`) for deterministic output
3. With high temperature (`1.0`) for more variation

Run the next cell multiple times to observe consistency vs creativity.

```python
add_user_message(
    messages,
    "Give me one short movie idea about an AI assistant and a detective."
)

without_temperature = chat(messages)
with_low_temperature = chat(messages, temperature=0.0)
with_high_temperature = chat(messages, temperature=1.0)

print("WITHOUT temperature (API default):\n")
print(without_temperature)
print("\n" + "=" * 80 + "\n")

print("WITH low temperature (0.0):\n")
print(with_low_temperature)
print("\n" + "=" * 80 + "\n")

print("WITH high temperature (1.0):\n")
print(with_high_temperature)
```

## How To Read the Results

- The `0.0` run should usually be the most stable and repeatable.
- The `1.0` run should show higher variety in wording and ideas.
- The no-temperature run uses model defaults and may land between those behaviors.
