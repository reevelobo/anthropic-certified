# Multi Conversation 

```python
# Multi Conversation 

def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)
    return messages

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    return messages.append(assistant_message)

def chat(messages):
    message = client.messages.create(
        model = model,
        max_tokens = 1000,
        messages = messages,
    )
    return message.content[0].text
```

#### Makes a starting list of messages

```python
messages = []
```

#### Add in the initial user question of "Defin quantum computing in one sentance"

```python
add_user_message(messages, "Define qunatum computing in one sentance")
```

#### Pass the list of messages into 'chat' to get an answer

```python
answer = chat(messages)

answer
```

#### Take the answer and add it as an assistant message into out list 

```python
add_assistant_message(messages, answer)
```

#### Add in the user's follow-up question

```python
add_user_message(messages, "Write an other sentance")
```

#### Call chat again with the list of messages to get a final answer

```python
answer = chat(messages)
answer
```
