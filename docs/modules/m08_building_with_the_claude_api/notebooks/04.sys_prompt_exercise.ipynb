```python
# Import

from anthropic import Anthropic
from dotenv import load_dotenv

# Env var
load_dotenv()
```

## Step 1: Imports and Environment

This cell imports the Anthropic SDK and loads environment variables.

Before running the notebook, make sure your API key is available in your runtime environment so the client can authenticate successfully.

```python
# Define modle and client
model = "claude-haiku-4-5-20251001"

client = Anthropic()
```

## Step 2: Configure Model and Client

This cell sets the model and creates the API client.

Tip: keep model assignment in one place so you can quickly test behavior changes across models later.

```python
## Make an initial list of messages
messages = []
```

## Step 3: Initialize Conversation State

The `messages` list stores the running conversation.

Each turn appends either a user message or an assistant message, allowing multi-turn context to accumulate naturally.

```python
## Define helper fucntions 
def add_user_message(messages, text):
    return messages.append({"role": "user", "content": text})

def add_assistant_message(messages, text):
    return messages.append({"role": "assistant", "content": text})

def chat(messages, system=None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
    }

    # Only include system when present.
    if system:
        params["system"] = system

    message = client.messages.create(**params)
    return message.content[0].text
```

## Step 4: Define Helper Functions

You now have three helpers:
- add_user_message: appends user turns
- add_assistant_message: appends assistant turns
- chat: sends the conversation to Claude

Notice the optional system argument in `chat(messages, system=None)`. This is the key extension that enables behavioral control.

```python
add_user_message(
    messages,
    "write a python function that checks a string for duplication."
)

answer = chat(messages)
```

## Step 5: Send a Baseline Prompt (No System Prompt)

This run captures the default behavior of the model for the same user request.

Use it as your control output before adding constraints.

```python
answer
```

## Step 6: Review the Baseline Output

Observe the default response style:
- Is it verbose?
- Does it include explanation text?
- Does it strictly return only code?

You will compare this directly with the constrained output in the next steps.

```python
system_prompt="""
You are a Python engineer who witer very concise code.
"""
answer = chat(messages, system=system_prompt)
```

## Step 7: Apply a System Prompt Constraint

This cell adds explicit rules:
- keep output minimal
- return only Python code

System prompts are useful when output format and response style must be consistent across requests.

```python
answer
```
