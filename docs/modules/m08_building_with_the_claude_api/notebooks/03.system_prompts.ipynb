# System Prompts

System prompts let you shape how Claude responds so behavior is consistent with your use case.

In this notebook, you will:
- Understand why system prompts matter
- See the difference in outputs with and without a system prompt
- Build a reusable chat function that accepts an optional system prompt

```python
import os
from anthropic import Anthropic

client = Anthropic(api_key=os.getenv("ANTHROPIC_API_KEY"))
model = "claude-sonnet-4-5"

print("Anthropic client initialized")
```

## Why System Prompts Matter

Consider a math tutor chatbot.

When a student asks:

> How do I solve 5x + 2 = 3 for x?

A good tutor should:
1. Give hints before full solutions
2. Guide step by step
3. Encourage independent thinking

Without a system prompt, the model may provide direct answers immediately.

## How System Prompts Work

System prompts are plain strings that define desired behavior. You pass them to `client.messages.create(...)` using the `system` parameter.

Key benefits:
1. Role-specific behavior
2. Better consistency
3. Better control of tone and constraints

```python
messages = [
    {
        "role": "user",
        "content": "How do I solve 5x + 2 = 3 for x?"
    }
]

def chat(messages, system=None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
    }

    # Only include system when present.
    if system:
        params["system"] = system

    message = client.messages.create(**params)
    return message.content[0].text
```

## Seeing the Difference

First, run without a system prompt.
Then run with a math tutor system prompt and compare outputs.

```python
answer_without_system = chat(messages)
print(answer_without_system)
```

```python
system_prompt = """
You are a patient math tutor.
Do not directly answer a student's questions.
Guide them to a solution step by step.
"""

answer_with_system = chat(messages, system=system_prompt)
print(answer_with_system)
```

## Summary

System prompts help you move from generic responses to consistent, purpose-driven behavior.

They are essential for production AI apps where tone, role, and interaction style must remain reliable.
