# Chat Bot Exercise 

```python
# Load Env Variables
from dotenv import load_dotenv
load_dotenv()
```

```python
## Import Statements 
from anthropic import Anthropic

# Define model and client
client = Anthropic()
model = "claude-haiku-4-5-20251001"
```

```python
## Make an initial list of messages
messages = []
```

```python
## Define helper fucntions 
def add_user_message(messages, text):
    return messages.append({"role": "user", "content": text})

def add_assistant_message(messages, text):
    return messages.append({"role": "assistant", "content": text})

def chat(messages):
    message = client.messages.create(
        model= model,
        max_tokens=1000,
        messages=messages
    )
    return message.content[0].text
```

```python

## User a 'while true' to run the chatbot forever
while True:
    # Get user input 
    user_input = input("> ")
    print(">", user_input)

    # Add user input to the list of messages 
    add_user_message(messages, user_input)

    # Call Claude wit the 'chat' function
    answer = chat(messages) 

    # add generated text to the list of message 
    add_assistant_message(messages, answer)

    # print the generated text 
    print(f"----\n{answer}\n----")
```
